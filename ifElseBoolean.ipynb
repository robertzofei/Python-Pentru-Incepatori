## Tipul de date Boolean
Mai tineti minte ca in tipurile de date cu care am lucrat pana acum, adica cele pentru numere si siruri de caractere, noi puteam sa memoram orice numar sau orice text pe care ni le imagiam? Ei bine, acum o sa lucram cu un tip de date in care poti stoca doar doua valori: True sau False, adica adevarat sau fals. Doua lucruri importante: valorile acestea nu apar scrise intre ghilimele (cum erau string-urile), si incep mereu cu litera mare.

```python
a = True
b = False
print(a,b)
```

Poate ca va intrebati de unde vine aceasta denumire de Boolean. Acest tip de date este numit dupa matematicianul britanic George Boole. El a inventat algebra logica ce ii poarta numele si a pus bazele functionarii tuturor calculatoarelor si a limbajelor de programare din ziua de azi.

## Operatorii de Comparare (==,  !=,  <,  >)
| Operator | Ce înseamnă |
| :--- | :--- |
| `==` | Egal cu |
| `!=` | Nu e egal cu |
| `<` | Mai mic decât |
| `>` | Mai mare decât |
| `<=` | Mai mic sau egal cu |
| `>=` | Mai mare sau egal cu |

Haideti sa vedem cateva exemple:

```python
1 == 1
```

```python
4 == 32
```

```python
2 != 3
```

```python
2 != 2
```

```python
64 == 64.0
```

```python
'salut' == 'salut'
```

```python
'salut' == 'Salut'
```

```python
True == True
```

```python
True != False
```
